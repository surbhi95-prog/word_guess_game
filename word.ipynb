```python
import random
import nltk
nltk.download('words') # corpus in NLTK that has English words

# this gives us a list of ~236000 english words
from nltk.corpus import words
word_list = words.words() # words is a corpus object and words() is a method that actually returns the words
```

```python
import random;
## CREATING A WORD BANK
print("\n+------------------+\n|  WORD GUESS GAME |\n+------------------+")
word_bank = random.sample(word_list,10) # taking sample of 20 words
word= random.choice(word_bank).lower()
score = 0
# EMPTY ARRAY with PLACEHOLDERS
empty_arr = []
for i in range(len(word)):
    empty_arr.append("_")

attempts = 8

while attempts > 0:

    print("\nCurrent word:"+' '.join(empty_arr))
    if ''.join(empty_arr) == word:
        print("\n------------You WON--------------")
        print("SCORE:",score)
        break
    else:
        letter = input("Guess a letter:")
        print("Guessed letter:",letter)
        if letter in word and letter not in empty_arr:
            print("Guessed Correctly!")
            for i in range(len(word)):
                if word[i] == letter:
                    empty_arr[i] = letter
                    score += 100/round(len(word),2)
        else:
            attempts -= 1
            score -=5
            print("Wrong Guess, ",attempts,"attempts left!")
if attempts == 0:
    print("\n------------x You Lose! x------------")
    print("\nThe word was:",word)

```
